```python
import shutil
import os

# Pick just 2 dt partitions — keep it light
partitions_to_export = ["dt=2024-01-02","dt=2024-01-03","dt=2024-01-04","dt=2024-01-05","dt=2024-01-06","dt=2024-01-07","dt=2024-01-08","dt=2024-01-09"]

for table in ["t_Loan", "t_LoanInstallment"]:
    src_base = f"/Volumes/bfsi_lakehouse/raw/synthetic_data/{table}"
    final_zip_path = f"/Volumes/bfsi_lakehouse/default/docs/{table}_export.zip"
    
    # Copy selected dt partitions to a temp local-style folder first
    temp_dir = f"/tmp/{table}_export"
    os.makedirs(temp_dir, exist_ok=True)
    
    for dt in partitions_to_export:
        src = f"{src_base}/{dt}"
        dst = f"{temp_dir}/{dt}"
        shutil.copytree(src, dst, dirs_exist_ok=True)
    
    # Zip it in /tmp first
    temp_zip = f"/tmp/{table}_export"
    shutil.make_archive(temp_zip, 'zip', temp_dir)
    
    # Copy to Volumes
    shutil.copy(f"{temp_zip}.zip", final_zip_path)
    print(f"✓ Created: {final_zip_path} (contains all selected dt partitions)")
```
